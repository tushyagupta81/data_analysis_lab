1.Create a function to check whether numpy array has Zero values or not.

```python
import numpy as np
```

```python
arr = np.random.randint(0, 100, 1000)
```

```python
def check_zeros(x) -> bool:
    for ele in x:
        if ele == 0:
            return True

    return False
```

```python
print(check_zeros(arr))
print(arr.any(0))
```

2.Consider an array contains marks of students. Find the median and mean of students who got passing marks.

```python
import numpy as np
```

```python
marks = np.random.rand(100) * 100
print(marks[:10])
```

```python
passing_marks = 33
passing_args = np.where(marks > passing_marks)
print(f"Passing indexes = {passing_args}")
print(f"Mean = {marks[passing_args].mean()}")
print(f"Median = {np.median(marks[passing_args])}")
```

3.Consider an array containing date of birth of students (Use np.datetime64('2018-02-01')). Find the age of all students. If age is 18 label it First, if it is 19, label it Second, if it is 20, label it Third, if it is 21, label it Fourth.

```python
import numpy as np
```

```python
base = np.random.randint(-5200, -3700, 100)
date = np.datetime64("2018-02-01")
birthdays = date + base
```

```python
print(birthdays.shape)
print(birthdays[:10])
```

```python
years_old = (np.datetime64("2026-02-04") - birthdays) / 365
years_old = years_old.reshape((years_old.size, 1))
```

```python
labels = np.array(["First", "Second", "Third", "Forth", "Fifth"])
labeled_rows = labels[np.array(years_old - 18, dtype=np.int32)]
```

```python
print(np.hstack((years_old.astype(np.int32), labeled_rows)))
```

4.Create a function to sort 2d Array by a column.

```python
import numpy as np
```

```python
def sort_col(x, col: int):
    return x[x[:, col].argsort()]
```

```python
arr = np.random.randint(0, 100, (3, 4))
print(arr)
print(sort_col(arr, 1))
```
