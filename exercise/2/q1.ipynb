1.Create a function to check whether numpy array has Zero values or not.

```python
import numpy as np
```

```python
arr = np.random.randint(0, 100, 1000)
```

```python
def check_zeros(x) -> bool:
    for ele in x:
        if ele == 0:
            return True

    return False
```

```python
print(check_zeros(arr))
print(arr.any(0))
```
