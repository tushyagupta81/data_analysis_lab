3.Consider an array containing date of birth of students (Use np.datetime64('2018-02-01')). Find the age of all students. If age is 18 label it First, if it is 19, label it Second, if it is 20, label it Third, if it is 21, label it Fourth.

```python
import numpy as np
```

```python
base = np.random.randint(-5200, -3700, 100)
date = np.datetime64("2018-02-01")
birthdays = date + base
```

```python
print(birthdays.shape)
print(birthdays[:10])
```

```python
years_old = (np.datetime64("2026-02-04") - birthdays) / 365
years_old = years_old.reshape((years_old.size, 1))
```

```python
labels = np.array(["First", "Second", "Third", "Forth", "Fifth"])
labeled_rows = labels[np.array(years_old - 18, dtype=np.int32)]
```

```python
print(np.hstack((years_old.astype(np.int32), labeled_rows)))
```
