### Exercise 10

Compare performance of logistic regression classifier on Iris, and digits dataset of sklearn by training set size from 1% to 100%. Plot the accuracy vs training set size graph.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_digits, load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
def find_and_plot(dataset, min_train=1, max_train=100, step=1):
    acc = []
    train_size = []
    for i in range(min_train, max_train, step):
        X_train, X_test, y_train, y_test = train_test_split(
            dataset.data, dataset.target, test_size=(1 - i / 100), random_state=42
        )
        lin = LogisticRegression(max_iter=200)
        lin.fit(X_train, y_train)
        y_pred = lin.predict(X_test)
        acc.append(accuracy_score(y_test, y_pred) * 100)
        train_size.append(i)

    plt.plot(train_size, acc)
    plt.xlabel("Train set percent")
    plt.ylabel("Accuracy")
    plt.ylim(0, 102)
    plt.xlim(0, 100)
    plt.grid()
    plt.show()
```

```python
find_and_plot(load_digits())
```

```python
find_and_plot(load_iris(), min_train=2)
```
