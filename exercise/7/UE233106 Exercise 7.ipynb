### Exercise 7

Read the Colleges dataset and clean all the numeric attributes.\
From all marks based attributes, remove % sign from different attributes. In SAT/ACT column, Keep higher value.\
From all cost based attributes, Remove $ sign and commas.\
Remove any null value in any attribute.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../../datasets/Top 100 Private Colleges.2003.csv")
df.set_index("Overall Rank", inplace=True)
df.head(10)
```

```python
df.isnull().sum() / len(df) * 100
```

```python
df.dtypes
```

```python
to_drop = []
cols = df.columns

na_ratio = df.isnull().sum() / len(df) * 100

for i in range(len(cols)):
    if na_ratio.values[i] >= 20:
        to_drop.append(cols[i])

print(to_drop)
```

```python
df.drop(to_drop, axis=1, inplace=True)
print(list(c for c in df.columns if c in to_drop))
```

```python
has_some_missing = []
cols = df.columns

na_ratio = df.isnull().sum() / len(df) * 100

for i in range(len(cols)):
    if na_ratio.values[i] >= 0:
        has_some_missing.append(cols[i])

print(has_some_missing)
```

```python
new_df = df.dropna(axis=0)
print(f"{len(new_df)}/{len(df)}")
```

```python
new_df.head()
```

```python
new_df.isnull().sum() / len(new_df) * 100
```

```python
col_with_percent = [
    "Admission Rate",
    "4-year Grad. Rate",
    "6-year Grad. Rate",
    "Need Met",
    "Aid From Grants",
]


def clean_percent(inp):
    if "%" in inp:
        inp = inp[: inp.find("%")]

    if inp.lower() == "na":
        return np.nan
    return inp


for col in col_with_percent:
    new_df[col] = new_df[col].apply(clean_percent).astype(np.float64)

new_df.head()
```

```python
new_df.dtypes
```

```python
unwanted = ["%", "*"]


def remove_unwanted(inp):
    for uw in unwanted:
        inp = inp.replace(uw, "")
    return inp


def clean_sat_act(inp):
    if "/" in inp:
        sat, act = inp.split("/")
        act = float(remove_unwanted(act))
        sat = float(remove_unwanted(sat))
        return max(sat, act)
    else:
        return float(remove_unwanted(inp))


new_df["*SAT or ACT"] = new_df["*SAT or ACT"].apply(clean_sat_act).astype(np.float64)
```

```python
new_df.head()
```

```python
col_with_dollar = [
    "Total Costs",
    "Cost After Need-based Aid",
    "Cost After Non-Need-Based Aid",
    "Average Debt",
]

unwanted_dollar = [",", "$"]


def clean_dollar(inp):
    for uw in unwanted_dollar:
        inp = inp.replace(uw, "")
    return inp


for col in col_with_dollar:
    new_df[col] = new_df[col].apply(clean_dollar).astype(np.float64)

new_df.head()
```
